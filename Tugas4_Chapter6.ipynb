## K-Nearest Neighbors (k-NN)

K-Nearest Neighbors (k-NN) adalah algoritma berbasis instans non-parametrik yang mengklasifikasikan suatu titik data baru berdasarkan mayoritas label dari tetangga terdekatnya. Kedekatan antar-titik data dihitung menggunakan fungsi jarak geometris, seperti Jarak Euclidean. Karena performa k-NN sangat bergantung pada skala metrik jarak, prapemrosesan berupa standardisasi atau normalisasi fitur wajib dilakukan sebelum melatih model agar fitur dengan rentang nilai besar tidak mendominasi perhitungan jarak.

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

data = {
    'X1': [1, 2, 8, 9],
    'X2': [2, 1, 9, 7],
    'Y': [0, 0, 1, 1]
}
df = pd.DataFrame(data)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[['X1', 'X2']])
y_vector = df['Y']

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X_scaled, y_vector)

print(model.predict(X_scaled))
```

## Decision Trees and Random Forests

Pohon Keputusan (Decision Tree) bekerja dengan cara mempartisi dataset secara rekursif menjadi subset yang lebih kecil berdasarkan aturan keputusan pada fitur-fiturnya. Parameter seperti Gini Impurity atau Entropy digunakan untuk mengukur tingkat kemurnian node hasil pemisahan data. Untuk mengatasi kelemahan Decision Tree tunggal yang sangat rentan mengalami overfitting, algoritma Random Forest menggabungkan puluhan hingga ratusan pohon keputusan independen menggunakan metode Bagging (Bootstrap Aggregating). Hasil prediksi akhir ditentukan melalui mekanisme pemungutan suara terbanyak (majority voting) untuk klasifikasi atau rata-rata untuk regresi.

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

data = {
    'X1': [1, 2, 3, 4, 5, 6],
    'X2': [6, 5, 4, 3, 2, 1],
    'Y': [0, 0, 0, 1, 1, 1]
}
df = pd.DataFrame(data)

X_matrix = df[['X1', 'X2']]
y_vector = df['Y']

model = RandomForestClassifier(n_estimators=10, random_state=42)
model.fit(X_matrix, y_vector)

print(model.feature_importances_)
```
