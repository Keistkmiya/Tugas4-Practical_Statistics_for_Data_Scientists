# Chapter 2: Data and Sampling Distributions

## Sampling Distribution and Central Limit Theorem (CLT)

Distribusi sampling adalah distribusi peluang dari sebuah statistik (misalnya rerata) yang diperoleh dari sejumlah besar sampel acak yang diambil dari populasi yang sama. Teorema Limit Pusat (Central Limit Theorem) menyatakan bahwa jika ukuran sampel cukup besar (biasanya $n \ge 30$), maka distribusi dari rerata sampel akan mendekati Distribusi Normal, tidak peduli apa pun bentuk distribusi populasi aslinya. Sifat ini menjadi fondasi penting dalam statistika inferensial karena memungkinkan kita untuk membuat estimasi dan uji hipotesis tanpa harus mengetahui bentuk distribusi populasi secara mendetail.

```python
import pandas as pd
import numpy as np

population = np.random.exponential(scale=2, size=10000)
sample_means = []

for _ in range(1000):
    sample = np.random.choice(population, size=40)
    sample_means.append(sample.mean())

df_means = pd.DataFrame(sample_means, columns=['SampleMean'])
print(df_means.head())
```

## Standard Error

Galat Baku (Standard Error atau SE) mengukur seberapa besar variabilitas atau fluktuasi nilai statistik sampel (seperti rerata sampel) antar-sampel yang berbeda terhadap nilai parameter populasi yang sebenarnya. Perlu dibedakan antara standar deviasi dan standar error; standar deviasi mengukur sebaran data pada tingkat individu dalam satu sampel, sedangkan standar error mengukur sebaran dari statistik sampel itu sendiri. Nilai standar error dihitung dengan membagi standar deviasi sampel dengan akar kuadrat dari ukuran sampel ($SE = s / \sqrt{n}$), yang berarti semakin besar ukuran sampel, semakin kecil nilai galat bakunya.

```python
import pandas as pd
import numpy as np
from scipy import stats

data = [23, 29, 24, 32, 21, 26, 25, 30]
df = pd.DataFrame(data, columns=['Value'])

std_error = stats.sem(df['Value'])
print(std_error)
```

## The Bootstrap

Bootstrap adalah metode non-parametrik yang sangat kuat untuk mengestimasi distribusi sampling dari suatu statistik dengan cara melakukan pengambilan sampel ulang (resampling) secara berulang-ulang dari dataset asli dengan pengembalian (with replacement). Ukuran setiap sampel bootstrap harus sama persis dengan ukuran dataset asli. Metode ini sangat diandalkan dalam data science karena tidak bergantung pada asumsi matematika kaku mengenai distribusi data (seperti asumsi bahwa data harus berdistribusi normal) dan sangat efektif digunakan pada dataset berukuran kecil atau pada statistik yang kompleks.

```python
import pandas as pd
import numpy as np

data = [15, 22, 19, 24, 18, 30, 27]
df = pd.DataFrame(data, columns=['Value'])

bootstrap_means = []
for _ in range(1000):
    boot_sample = df['Value'].sample(n=len(df), replace=True)
    bootstrap_means.append(boot_sample.mean())

df_boot = pd.DataFrame(bootstrap_means, columns=['BootMean'])
print(df_boot.head())
```

## Confidence Intervals

Selang Kepercayaan (Confidence Interval atau CI) adalah rentang nilai yang digunakan untuk mengestimasi parameter populasi yang tidak diketahui (seperti rerata populasi) berdasarkan statistik sampel. Rentang ini disertai dengan tingkat kepercayaan tertentu (misalnya 95%). Tingkat kepercayaan 95% berarti jika kita mengambil sampel secara acak berulang kali dan menghitung selang kepercayaan dari masing-masing sampel tersebut, maka 95% dari selang-selang kepercayaan tersebut akan mencakup nilai parameter populasi yang sebenarnya. Batas atas dan batas bawah selang ini dapat dihitung menggunakan distribusi sampling dari hasil metode bootstrap.

```python
import pandas as pd
import numpy as np

data = [15, 22, 19, 24, 18, 30, 27]
df = pd.DataFrame(data, columns=['Value'])

bootstrap_means = []
for _ in range(1000):
    boot_sample = df['Value'].sample(n=len(df), replace=True)
    bootstrap_means.append(boot_sample.mean())

lower_bound = np.percentile(bootstrap_means, 2.5)
upper_bound = np.percentile(bootstrap_means, 97.5)

print(lower_bound)
print(upper_bound)
```

## Normal Distribution vs. Long-Tailed Distributions

Distribusi Normal adalah distribusi kontinu berbentuk lonceng simetris sempurna di mana nilai rerata, median, dan modus berada tepat di satu titik pusat yang sama. Namun, dataset dunia nyata dalam bidang data science sering kali tidak simetris dan mengikuti Distribusi Ekor Panjang (Long-Tailed Distributions) atau condong (skewed). Pada distribusi ekor panjang, sebagian besar data berkumpul di satu sisi, tetapi terdapat ekor panjang berisi nilai ekstrem yang menjulur ke sisi lainnya. Memahami bentuk distribusi ini sangat penting karena banyak model machine learning berasumsi data berdistribusi normal, sehingga transformasi data (seperti log transformation) sering diperlukan jika data memiliki ekor yang panjang.

```python
import numpy as np
import scipy.stats as stats

normal_data = np.random.normal(loc=0, scale=1, size=1000)
skewed_data = stats.skewnorm.rvs(a=10, size=1000)

print(np.mean(normal_data))
print(stats.skew(skewed_data))
```

## Binomial and Poisson Distributions

Distribusi Binomial dan Distribusi Poisson adalah dua jenis distribusi peluang diskrit yang mengukur frekuensi kejadian tertentu. Distribusi Binomial digunakan untuk memodelkan jumlah keberhasilan dalam sejumlah percobaan independen yang bersifat dikotomi (hanya ada dua kemungkinan hasil: sukses atau gagal, seperti konversi klik iklan). Sementara itu, Distribusi Poisson digunakan untuk memodelkan berapa kali suatu peristiwa acak terjadi dalam interval waktu atau ruang yang ditentukan (seperti jumlah kunjungan server per menit). Kedua distribusi ini merupakan fondasi penting dalam menganalisis data kategorikal dan pemodelan prediktif berbasis probabilitas.

```python
import scipy.stats as stats

binomial_prob = stats.binom.pmf(k=3, n=5, p=0.5)
poisson_prob = stats.poisson.pmf(k=2, mu=3)

print(binomial_prob)
print(poisson_prob)
```
