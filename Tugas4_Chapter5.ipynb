## Simple Linear Regression

Regresi Linear Sederhana adalah model statistik yang digunakan untuk memodelkan hubungan linear antara satu variabel prediktor independen ($X$) dan satu variabel respon dependen ($Y$). Persamaan garis regresi ini dinyatakan sebagai $Y = b_0 + b_1X$, di mana $b_0$ adalah intersep (titik potong sumbu $Y$) dan $b_1$ adalah kemiringan garis (slope/koefisien). Tujuan utama dari model ini adalah menemukan garis lurus terbaik yang meminimalkan jumlah kuadrat dari sisaan (residuals) atau jarak vertikal antara titik data asli dengan garis prediksi tersebut.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'X': [1.5, 2.0, 3.0, 4.5, 5.0],
    'Y': [2.5, 3.1, 4.8, 6.2, 7.0]
}
df = pd.DataFrame(data)

X_matrix = df[['X']]
y_vector = df['Y']

model = LinearRegression()
model.fit(X_matrix, y_vector)

print(model.intercept_)
print(model.coef_[0])
```

## Multiple Linear Regression and R-Squared

Regresi Linear Berganda perluas fungsi dari regresi sederhana dengan melibatkan dua atau lebih variabel independen untuk memprediksi satu variabel dependen. Untuk mengukur seberapa baik model ini dalam menjelaskan variabilitas data, digunakan metrik R-Squared ($R^2$) atau koefisien determinasi. Nilai $R^2$ berkisar antara 0 hingga 1, di mana nilai yang mendekati 1 menandakan bahwa sebagian besar variasi dari variabel target berhasil dijelaskan oleh kombinasi variabel-variabel prediktor yang dimasukkan ke dalam model.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'X1': [1, 2, 3, 4, 5],
    'X2': [5, 4, 2, 1, 1],
    'Y': [15, 19, 21, 26, 30]
}
df = pd.DataFrame(data)

X_matrix = df[['X1', 'X2']]
y_vector = df['Y']

model = LinearRegression()
model.fit(X_matrix, y_vector)

r_squared = model.score(X_matrix, y_vector)

print(model.intercept_)
print(model.coef_)
print(r_squared)
```

## Regression Diagnostics and Residual Analysis

Diagnostik regresi digunakan untuk mengevaluasi apakah asumsi-asumsi dasar dari model regresi linear terpenuhi setelah model tersebut dilatih. Alat utama dalam melakukan diagnosis ini adalah analisis sisaan (Residual Analysis), yaitu selisih antara nilai aktual dengan nilai prediksi ($e = y - \hat{y}$). Sisaan yang ideal harus berdistribusi normal, memiliki varians yang konstan di seluruh rentang prediksi (Homoskedastisitas), dan tidak menunjukkan pola geometris tertentu jika diplot terhadap nilai prediksi. Jika plot sisaan membentuk pola corong atau kurva, hal tersebut mengindikasikan adanya masalah Heteroskedastisitas atau ketidaklinearan data.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'X': [1, 2, 3, 4, 5, 6, 7, 8],
    'Y': [3, 5, 6, 8, 11, 13, 14, 17]
}
df = pd.DataFrame(data)

X_matrix = df[['X']]
y_vector = df['Y']

model = LinearRegression()
model.fit(X_matrix, y_vector)

predictions = model.predict(X_matrix)
residuals = y_vector - predictions

print(residuals.tolist())
```

## Confounding Variables and Multicollinearity

Dalam pemodelan statistik berganda, ketepatan interpretasi koefisien regresi sering kali terganggu oleh dua fenomena: Confounding Variables dan Multikolinieritas. Confounding terjadi ketika ada variabel luar yang memengaruhi variabel independen sekaligus variabel dependen, sehingga memicu hubungan semu. Sementara itu, Multikolinieritas terjadi ketika sesama variabel prediktor (independen) memiliki korelasi linear yang sangat kuat satu sama lain. Kondisi ini membuat estimasi koefisien regresi menjadi tidak stabil dan sulit untuk mengisolasi pengaruh murni dari masing-masing fitur terhadap target.

```python
import pandas as pd
import numpy as np

data = {
    'X1': [1, 2, 3, 4, 5],
    'X2': [2.1, 3.9, 6.1, 7.9, 10.1],
    'Y': [10, 14, 19, 22, 26]
}
df = pd.DataFrame(data)

matrix_corr = df[['X1', 'X2']].corr()
print(matrix_corr)
```

## Outliers and Influential Points (Cook's Distance)

Dalam pemodelan regresi, tidak semua pencilan (outliers) memiliki dampak yang sama terhadap garis prediksi. Outlier adalah titik data yang nilai aktualnya sangat jauh dari nilai prediksi model (memiliki residual besar). Namun, jika titik data tersebut juga memiliki nilai prediktor ($X$) yang ekstrem, titik tersebut dinamakan *Influential Point* karena memiliki daya ungkit besar (*high leverage*) untuk menarik garis regresi ke arahnya. Jarak Cook (*Cook's Distance*) adalah metrik statistik yang digunakan untuk mengukur seberapa besar perubahan pada seluruh koefisien regresi jika suatu titik data tertentu dihapus dari dataset. Nilai Jarak Cook yang tinggi menandakan bahwa poin tersebut sangat mendominasi arah model.

```python
import numpy as np
import statsmodels.api as sm
import pandas as pd

X = np.array([1, 2, 3, 4, 5, 20])
y = np.array([2, 4, 5, 4, 6, 25])

X_with_constant = sm.add_constant(X)
model = sm.OLS(y, X_with_constant).fit()

influence = model.get_influence()
cooks_dist = influence.cooks_distance[0]

print(cooks_dist.tolist())
```

## Polynomial and Spline Regression

Ketika hubungan antara variabel independen dan dependen tidak membentuk garis lurus, model regresi linear sederhana akan menghasilkan bias yang tinggi. Untuk menangani tren non-linear, kita bisa menggunakan Regresi Polinomial dengan menambahkan pangkat dari variabel prediktor (seperti $X^2, X^3$) ke dalam persamaan matriks desain. Namun, regresi polinomial derajat tinggi berisiko mengalami osilasi ekstrem di ujung data. Sebagai alternatif yang lebih stabil, Regresi Spline membagi data menjadi beberapa interval berdasarkan titik simpul (*knots*) dan mencocokkan fungsi polinomial tingkat rendah yang terpisah di setiap interval tersebut secara mulus (*smooth*).

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([1, 4, 9, 15, 24])

poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

print(model.intercept_)
print(model.coef_)
```
