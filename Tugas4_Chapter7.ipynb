## K-Means Clustering

K-Means Clustering adalah algoritma pembelajaran tidak terawasi yang mempartisi dataset ke dalam sejumlah $K$ kluster yang terpisah. Algoritma ini bekerja secara iteratif dengan meminimalkan jumlah kuadrat jarak (biasanya jarak Euclidean) antara titik data dengan pusat kluster (centroid) terdekatnya. Proses penentuan jumlah kluster $K$ yang optimal tidak didasarkan pada label data, melainkan dianalisis secara statistik menggunakan metode visualisasi seperti Elbow Method atau nilai Silhouette Score untuk melihat titik di mana penurunan varians dalam kluster mulai melambat.

```python
import pandas as pd
from sklearn.cluster import KMeans

data = {
    'X1': [1, 1.5, 5, 6, 8, 8.5],
    'X2': [1, 2, 6, 5, 9, 8.5]
}
df = pd.DataFrame(data)

model = KMeans(n_clusters=2, random_state=42, n_init=10)
model.fit(df)

print(model.labels_)
print(model.cluster_centers_)
```

## Hierarchical Clustering

Berbeda dengan K-Means yang mengharuskan kita menentukan jumlah kluster sejak awal, Hierarchical Clustering membangun struktur kluster secara bertahap seperti silsilah pohon. Pendekatan yang paling umum adalah tipe aglomeratif (bottom-up), di mana setiap titik data dianggap sebagai kluster mandiri pada awalnya, lalu pasangan kluster yang paling berdekatan digabungkan satu per satu secara iteratif. Representasi grafis dari proses penggabungan ini disebut Dendrogram, yang sangat berguna bagi seorang analis untuk memotong pohon pada ketinggian tertentu guna menentukan jumlah kluster paling logis berdasarkan jarak kedekatan antar-grup.

```python
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram

data = {
    'X1': [1, 2, 5, 7],
    'X2': [2, 1, 6, 8]
}
df = pd.DataFrame(data)

Z = linkage(df, method='ward')
print(Z)
```
