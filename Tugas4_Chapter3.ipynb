## A/B Testing and Hypothesis Testing

Uji A/B (A/B Testing) adalah eksperimen statistik acak yang membandingkan dua versi perlakuan—biasanya versi kontrol (A) dan versi baru (B)—untuk menentukan versi mana yang memberikan kinerja lebih baik berdasarkan metrik tertentu. Proses ini diawali dengan merumuskan Hipotesis Nol ($H_0$), yang berasumsi bahwa tidak ada perbedaan nyata antar-kedua kelompok tersebut dan variasi yang terjadi hanya karena faktor kebetulan. Sebaliknya, Hipotesis Alternatif ($H_a$) menyatakan adanya perbedaan yang signifikan. Pengujian hipotesis dilakukan untuk mengumpulkan bukti statistik guna menolak atau gagal menolak hipotesis nol.

```python
import pandas as pd
import numpy as np

group_A = np.random.binomial(1, 0.10, 1000)
group_B = np.random.binomial(1, 0.12, 1000)

df_A = pd.DataFrame(group_A, columns=['Conversion'])
df_B = pd.DataFrame(group_B, columns=['Conversion'])

print(df_A['Conversion'].mean())
print(df_B['Conversion'].mean())
```

## Resampling Permutation Test

Uji permutasi (Permutation Test) adalah jenis resampling non-parametrik yang menggabungkan seluruh data dari kedua kelompok eksperimen (A dan B), kemudian mengacak dan membaginya kembali secara acak ke dalam kelompok-kelompok baru dengan ukuran yang sama. Proses ini diulangi ribuan kali untuk melihat seberapa sering perbedaan acak yang dihasilkan menyamai atau melebihi perbedaan nyata yang diamati pada eksperimen asli. Uji ini sangat efektif karena tidak bergantung pada asumsi matematis tertentu mengenai bentuk distribusi data.

```python
import pandas as pd
import numpy as np

data_A = [12, 15, 18, 20]
data_B = [14, 19, 22, 25]
all_data = np.array(data_A + data_B)

observed_diff = np.mean(data_B) - np.mean(data_A)
perm_diffs = []

for _ in range(1000):
    np.random.shuffle(all_data)
    new_A = all_data[:len(data_A)]
    new_B = all_data[len(data_A):]
    perm_diffs.append(np.mean(new_B) - np.mean(new_A))

p_value = np.mean(np.array(perm_diffs) >= observed_diff)
print(p_value)
```

## Statistical Significance and P-Values

Signifikansi statistik mengukur apakah perbedaan hasil yang diamati dalam sebuah eksperimen benar-benar terjadi karena perlakuan yang diberikan atau hanya sekadar variasi acak (kebetulan). Nilai $p$ ($p$-value) adalah probabilitas untuk mendapatkan hasil eksperimen yang sama ekstremnya atau lebih ekstrem dari hasil yang diamati, dengan asumsi bahwa Hipotesis Nol ($H_0$) benar. Jika nilai $p$ sangat kecil (biasanya kurang dari ambang batas $\alpha = 0.05$), kita menolak hipotesis nol dan menyimpulkan bahwa perbedaan tersebut signifikan secara statistik.

```python
import scipy.stats as stats

data_A = [12, 15, 18, 20, 22, 24]
data_B = [16, 20, 23, 25, 28, 30]

t_stat, p_val = stats.ttest_ind(data_A, data_B)

print(t_stat)
print(p_val)
```

## t-Tests

Uji-t (t-test) adalah uji statistik parametrik standar yang digunakan untuk membandingkan rata-rata dari dua kelompok data guna menentukan apakah keduanya berbeda secara signifikan. Uji ini didasarkan pada statistik-t, yang mengukur rasio antara perbedaan rata-rata kedua kelompok dengan variabilitas data (standar error). Uji-t mengasumsikan bahwa data numerik yang diuji berdistribusi normal atau mendekati normal. Di dalam pustaka SciPy Python, operasi ini difasilitasi oleh fungsi `stats.ttest_ind()`.

```python
import scipy.stats as stats

group1 = [84, 88, 90, 92, 95]
group2 = [78, 80, 85, 86, 89]

t_stat, p_val = stats.ttest_ind(group1, group2)

print(t_stat)
print(p_val)
```

## ANOVA (Analysis of Variance)

Ketika kita perlu membandingkan rata-rata dari tiga atau lebih kelompok eksperimen secara bersamaan, uji-t konvensional tidak lagi efisien karena dapat meningkatkan risiko kesalahan tipe I (false positive). Sebagai solusinya, ANOVA (Analysis of Variance) digunakan untuk menguji apakah variabilitas antar-kelompok lebih besar dibandingkan variabilitas di dalam kelompok itu sendiri. Statistik uji yang digunakan dinamakan nilai $F$. Jika nilai $p$ dari hasil uji ANOVA lebih kecil dari tingkat signifikansi, hal ini menunjukkan bahwa minimal ada satu kelompok yang memiliki rata-rata berbeda secara signifikan dari kelompok lainnya.

```python
import scipy.stats as stats

method_A = [23, 25, 21, 24, 26]
method_B = [31, 29, 34, 32, 30]
method_C = [18, 20, 19, 22, 21]

f_stat, p_val = stats.f_oneway(method_A, method_B, method_C)

print(f_stat)
print(p_val)
```

## Chi-Square Test

Uji Chi-Square (Uji Kait-Kuadrat) digunakan khusus untuk menganalisis data kategorikal, terutama untuk menguji independensi atau hubungan antara dua variabel kategorikal dalam tabel kontingensi. Uji ini bekerja dengan cara membandingkan nilai frekuensi yang benar-benar diamati di lapangan (observed frequencies) dengan nilai frekuensi yang diharapkan secara teoretis (expected frequencies) jika diasumsikan kedua variabel tersebut tidak saling berhubungan. Jika perbedaan antara nilai pengamatan dan nilai harapan cukup besar hingga menghasilkan nilai $p$ yang signifikan, kita dapat menyimpulkan bahwa kedua variabel kategorikal tersebut saling berkaitan.

```python
import scipy.stats as stats
import numpy as np

observed = np.array([[20, 30], [30, 15]])
chi2_stat, p_val, dof, expected = stats.chi2_contingency(observed)

print(chi2_stat)
print(p_val)
```
